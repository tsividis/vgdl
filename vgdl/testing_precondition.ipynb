```python
class Precondition(object):
	def __init__(self, text, fn):
		self.text = text
		self.fn = fn

	def check(self, arg):
		return self.fn(arg) #1 is dummy arg

	def display(self):
		print self.text

test_p = Precondition('health>1', lambda x: x['health']>1)
b = {'health':2, 'books':1}
```

```python
test_p.check(b)
```
